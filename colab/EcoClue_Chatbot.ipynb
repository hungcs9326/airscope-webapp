## 作業重點說明

我沒有沿用老師的「員瑛式思考」人設，而是結合自己的環境工程背景，設計 **EcoClue 校園環境線索偵探**。

### 輸入

使用者會提供：

1. 現場文字描述，例如異味、顏色、位置、時間或已做過的處理。
2. 調查模式：空氣／異味、水質外觀、廢棄物、熱濕噪音或自由描述。
3. 回覆深度：快速判讀或標準調查。
4. 前面的對話紀錄。

### 輸出

機器人不會直接武斷下結論，而會：

- 先重述目前已知的「觀察事實」。
- 每回合只追問一個最有辨識力的問題。
- 線索足夠後列出 2～3 個可能方向，並清楚標示仍需確認。
- 提供低風險、可以立刻執行的下一步。
- 遇到疑似火災、瓦斯、化學品洩漏或明顯身體不適時，優先提醒離開現場並通報。

### 第二輪：微調 Prompt

> 不要做成百科型問答。請改成「現場線索偵探」：先整理使用者真的觀察到什麼，每回合只問一個最能縮小範圍的問題；資料足夠後才列出可能方向，而且要把事實與推測分開。它不能假裝做過量測，也不能做健康或法規判定。請再建議適合的 Gradio 互動元件。

### AI 第二輪回答

> 可以採用「重述已知線索 → 追問一題 → 提出可能方向 → 建議安全下一步」的調查流程。Gradio 可增加事件類型選單、回覆深度、範例按鈕與清除對話。若描述涉及刺鼻氣味、化學品、煙霧或身體不適，應先給安全提醒，再繼續詢問。

## 4. 最終 System Prompt：EcoClue 的完整人設

這一版不是只指定語氣，而是同時定義角色背景、調查流程、回覆格式、模式差異與安全界線。

> 📸 **重點截圖 2：**建議截下方 System Prompt 的部分程式碼，以及後面的單次 API 測試結果。


```python
BASE_SYSTEM_PROMPT = r"""
# 名稱與角色
你叫「阿澄」，是 EcoClue 校園環境線索偵探。
你像一位冷靜、有耐心、願意陪同學查線索的環工學長姐，但不能假裝自己是現場檢測人員、醫師或主管機關。

# 核心任務
使用者會描述校園、宿舍或生活空間中的環境現象。
你的任務不是立刻猜出唯一答案，而是把零散描述整理成可查證的線索，
透過持續對話協助使用者決定「下一個最值得確認的資訊」與「安全、低風險的下一步」。

# 調查流程
1. 先用一句話重述目前已知的觀察，不能把推測寫成事實。
2. 若線索不足，每回合只問一個最有辨識力的問題，不可一次列出問卷。
3. 線索較完整時，提出 2～3 個「可能方向」，使用「較可能、也可能、仍需確認」等不確定語氣。
4. 每個可能方向都要連回使用者提供的線索，不能憑空添加量測值、來源、時間或人物。
5. 最後提供 1～3 個低風險、可執行的下一步，例如記錄時間、比較位置、停止接觸或通知管理單位。
6. 後續對話要記得前文，不重複問使用者已回答過的問題。

# 回覆版型
線索不足時使用：
🔎 已知線索：...
❓ 下一個關鍵問題：...（只能一題）

線索足夠時使用：
🔎 已知線索：...
🧭 可能方向：...（2～3 點）
✅ 安全下一步：...（1～3 點）
❓ 若要繼續調查：...（至多一題）

# 語氣
- 使用台灣繁體中文。
- 像學長姐一起查問題，具體、友善但不賣弄術語。
- 不責怪使用者，不製造恐慌。
- 預設控制在 220 個中文字內；需要說明時才略為增加。

# 絕對限制
- 不可捏造感測器讀值、檢驗結果、法規門檻、污染來源或引用資料。
- 不做醫療診斷、法律判定或「一定安全／一定超標」的保證。
- 不把氣味直接等同特定化學物質。
- 不要求使用者聞、摸、混合、加熱或品嘗未知物質。
- 若出現煙霧、火花、疑似瓦斯／化學品洩漏、呼吸困難、暈眩、灼痛等急迫訊號，
  先請使用者停止接觸、離開現場並通知校方或緊急單位，不要繼續自行調查。
"""

MODE_RULES = {
    "空氣／異味": "優先確認時間、位置、氣味分布、通風或鄰近活動；不要由氣味名稱直接判定物質。",
    "水質外觀": "優先確認顏色、混濁、漂浮物、出現位置與冷熱水差異；不可建議直接飲用或品嘗。",
    "廢棄物": "優先確認物品材質、是否沾污、是否含電池／尖銳物／化學品與校內回收規則。",
    "熱濕／噪音": "優先確認時間、持續性、位置差異、活動來源與是否有可用量測資料。",
    "自由描述": "先判斷屬於哪類環境現象，再選擇最有辨識力的一個問題。"
}

DEPTH_RULES = {
    "快速判讀": "回覆要非常精簡，原則上 120～180 個中文字。",
    "標準調查": "可完整呈現推理線索與下一步，原則上 180～300 個中文字。"
}

def build_system_prompt(mode="自由描述", depth="標準調查"):
    mode_rule = MODE_RULES.get(mode, MODE_RULES["自由描述"])
    depth_rule = DEPTH_RULES.get(depth, DEPTH_RULES["標準調查"])
    return (
        BASE_SYSTEM_PROMPT
        + f"\n# 本次調查模式\n{mode}：{mode_rule}"
        + f"\n# 本次回覆深度\n{depth_rule}"
    )
```

## 5. 建立模型與先做單次測試

這裡沿用老師的 AISuite 呼叫方式，供應商改用 Groq。測試題故意只提供部分線索，確認機器人會先整理事實並追問，而不是直接武斷判定污染來源。


```python
MODEL = "groq:openai/gpt-oss-120b"
client = ai.Client()

def ask_once(message, mode="自由描述", depth="標準調查"):
    messages = [
        {"role": "system", "content": build_system_prompt(mode, depth)},
        {"role": "user", "content": message.strip()}
    ]

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=0.35,
        max_tokens=600
    )
    return response.choices[0].message.content
```

```python
# 單次測試：執行後請確認回答有「已知線索」與「只問一個問題」
test_question = "教室下午突然有一股像油漆的味道，開窗十分鐘後還是聞得到。"
test_answer = ask_once(
    test_question,
    mode="空氣／異味",
    depth="標準調查"
)

print("【測試輸入】")
print(test_question)
print("\n【EcoClue 回覆】")
print(test_answer)
```

### 我判斷測試是否成功的標準

- 有重述「下午、教室、類似油漆味、開窗後仍存在」等已提供線索。
- 沒有直接宣稱是哪一種化學物質或一定來自施工。
- 一次只問一個能縮小範圍的問題。
- 若提到刺鼻、不適或洩漏，會把離開現場與通報放在優先位置。
- 回覆維持台灣繁體中文，長度不會像百科文章。

若第一次結果一次問了很多題，我會在 System Prompt 加強「只能保留最重要的一題」；若它把推測寫成結論，我會加強「每個可能方向都必須使用不確定語氣」。


## 6. 用 Gradio 製作可持續對話的 Web App

Gradio 會把先前的 user／assistant 訊息一起交給模型，因此 EcoClue 可以記得使用者已提供的線索。傳給 API 前會清理 Gradio 的額外欄位，避免不同版本的訊息格式造成錯誤。


```python
def clean_message(item):
    """把 Gradio 訊息轉成 Groq／OpenAI 相容的 role-content 格式。"""
    if isinstance(item, dict):
        return {
            "role": item.get("role"),
            "content": str(item.get("content", ""))
        }

    return {
        "role": getattr(item, "role", ""),
        "content": str(getattr(item, "content", ""))
    }


def chat_fn(message, chat_history, mode, depth):
    chat_history = chat_history or []

    if not message or not message.strip():
        return "", chat_history

    api_history = []
    # 只保留最近 12 則 user／assistant 訊息，兼顧記憶與 API 用量
    for item in chat_history[-12:]:
        cleaned = clean_message(item)
        if cleaned["role"] in {"user", "assistant"} and cleaned["content"]:
            api_history.append(cleaned)

    messages = [
        {"role": "system", "content": build_system_prompt(mode, depth)},
        *api_history,
        {"role": "user", "content": message.strip()}
    ]

    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            temperature=0.35,
            max_tokens=700
        )
        answer = response.choices[0].message.content

    except Exception as error:
        # 不在畫面上顯示金鑰或完整系統資訊
        answer = (
            "⚠️ API 呼叫沒有成功。請確認 Colab Secrets 中的 Groq 金鑰、"
            "網路連線與模型額度後再試一次。"
        )
        print("API error:", type(error).__name__)

    updated_history = chat_history + [
        {"role": "user", "content": message.strip()},
        {"role": "assistant", "content": answer}
    ]

    return "", updated_history


def clear_chat():
    return "", []
```

```python
EXAMPLES = {
    "🏫 教室異味": "教室下午突然有一股像油漆的味道，開窗十分鐘後還是聞得到。",
    "🚰 飲水外觀": "宿舍飲水機剛裝的水看起來有一點白白的，放一下又比較透明。",
    "♻️ 回收判斷": "喝完的手搖飲杯裡還有珍珠和冰塊，我不知道校內要怎麼丟。"
}

custom_css = """
.gradio-container {
    max-width: 1080px !important;
    margin: 0 auto !important;
}
.eco-card {
    border: 1px solid #d5e8df;
    border-radius: 16px;
    padding: 14px;
    background: linear-gradient(135deg, #f3fbf6, #f3f8ff);
}
"""

with gr.Blocks(
    css=custom_css,
    theme=gr.themes.Soft(primary_hue="emerald", secondary_hue="sky")
) as demo:

    gr.Markdown("""
    # 🔎 EcoClue｜校園環境線索偵探
    **先整理觀察，再決定下一步。**  
    描述你在教室、宿舍或校園遇到的環境現象，阿澄會陪你逐步縮小方向。
    """)

    with gr.Row():
        with gr.Column(scale=1, min_width=250):
            gr.Markdown("""
            <div class="eco-card">
            <b>使用方式</b><br>
            ① 選擇調查模式<br>
            ② 選擇回覆深度<br>
            ③ 描述時間、位置與看到／聞到的現象<br><br>
            ⚠️ 本工具不能取代現場檢測、醫療或法規判定。
            </div>
            """)

            mode = gr.Dropdown(
                choices=list(MODE_RULES.keys()),
                value="自由描述",
                label="調查模式"
            )

            depth = gr.Radio(
                choices=list(DEPTH_RULES.keys()),
                value="標準調查",
                label="回覆深度"
            )

            clear = gr.Button("🧹 清除對話", variant="secondary")

        with gr.Column(scale=3):
            chatbot = gr.Chatbot(
                label="和阿澄一起查線索",
                height=470
            )

            with gr.Row():
                msg = gr.Textbox(
                    placeholder="例如：教室下午出現一股不熟悉的氣味……",
                    container=False,
                    scale=5
                )
                send = gr.Button("送出", variant="primary", scale=1)

    gr.Markdown("### 一鍵帶入測試情境")
    with gr.Row():
        air_btn = gr.Button("🏫 教室異味")
        water_btn = gr.Button("🚰 飲水外觀")
        waste_btn = gr.Button("♻️ 回收判斷")

    air_btn.click(lambda: EXAMPLES["🏫 教室異味"], outputs=msg)
    water_btn.click(lambda: EXAMPLES["🚰 飲水外觀"], outputs=msg)
    waste_btn.click(lambda: EXAMPLES["♻️ 回收判斷"], outputs=msg)

    msg.submit(
        fn=chat_fn,
        inputs=[msg, chatbot, mode, depth],
        outputs=[msg, chatbot]
    )

    send.click(
        fn=chat_fn,
        inputs=[msg, chatbot, mode, depth],
        outputs=[msg, chatbot]
    )

    clear.click(
        fn=clear_chat,
        inputs=None,
        outputs=[msg, chatbot]
    )
```

```python
# 執行後會出現 Gradio 介面與 share 連結
demo.queue().launch(share=True, debug=True)
```
